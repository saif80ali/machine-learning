```python
import pandas as pd
import pymysql
```

```python
connection = pymysql.connect(
    host="localhost",
    user="root",
    password="",
    database="ecommerce_training"
)
```

```python
connection
```

## To Run with Cursor and see the table results

```python
try:
    with connection.cursor() as cursor:
        cursor.execute("SELECT * FROM customers LIMIT 5")

        records = cursor.fetchall()

        for record in records:
            print(record)

finally:
    connection.close()
```

## To run SQL commands in pandas run this shell

```python
try:
    df = pd.read_sql("SELECT * FROM customers", connection)
    print(df.head())
finally:
    connection.close()
```
